```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-10,10,100)
```

```python
def plot_function(x, y, title, ylabel, derivative=None, d_label=None):
    plt.figure(figsize=(10,6))
    plt.plot(x, y, label=title, color="blue", linewidth=2)

    if derivative is not None:
        plt.plot(x, derivative, label=d_label, color="red", linestyle="--")

    plt.axhline(0, color="black", linewidth=0.5)
    plt.axvline(0, color="black", linewidth=0.5)

    plt.title(title)
    plt.xlabel("Input (x)")
    plt.ylabel(ylabel)

    plt.grid(True, which='both', linestyle="--", alpha=0.5)
    plt.legend()
    plt.show()
```

```python
def linear(x):
  return x

plot_function(x,linear(x),)
```

```python
def sigmoid(x):
  return 1 / (1+ np.exp(-x))
```

```python
def derivative_sigmoid(x):
  s = sigmoid(x)
  return s * (1-s)
```

```python
def tanh(x):
  return np.tanh(x)
```

```python
def derivative_tanh(x):
  return 1 - np.tanh(x)**2
```

```python
def relu(x):
  return np.maximum(0,x)
```

```python
def derivative_relu(x):
  r = relu(x)
  return np.where(x>0,1,0)
```

```python
def binary(x):
  return np.where(x>0,1,0)
```

```python
def swish(x):
  return x * sigmoid(x)
```

```python
plot_function(x,linear(x),"Linear Function","f(x)=x")
```

```python
plot_function(x,sigmoid(x),"Sigmoid Function","y",derivative_sigmoid(x),"Derivative")
```

```python
plot_function(x,tanh(x),"Tanh Function","y",derivative_tanh(x),"Dervative")
```

```python
plot_function(x,relu(x),"Relu","y",derivative_relu(x),"Derivative")
```

```python
plot_function(x,binary(x),"Binary","y")
```

```python
plot_function(x,swish(x),"Swish","y")
```
