```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('assignment_10_iris_dataset.csv')
```

```python
df['Species'] = df['Species'].astype('category')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten() # Flatten the 2D array to a 1D array for easy looping

columns = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']

for i, col in enumerate(columns):
    sns.histplot(df[col], bins=20, kde=True, ax=axes[i]) # Notice the ax=axes[i]
    axes[i].set_title(f"Distribution of {col}")

plt.tight_layout() # Prevents titles from overlapping
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=df)

plt.xlabel('Features')
plt.ylabel('Values')
plt.title("Box plot of all features")
```
