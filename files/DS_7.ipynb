```python
text =""" Football is the greatest sport out there nothing beats football the talent and determination is on another level plus additional loyalty by fans """
```

```python
print("OG : ",text)
```

```python
words = word_tokenize(text)

print(words)
```

```python
pos_tags = pos_tag(words)

print(pos_tags)
```

```python
stop_words = set(stopwords.words('english'))

words_no_sw = [w for w in words if w.lower() not in stop_words and w.isalpha()]

print(words_no_sw)
```

```python
stemmer = PorterStemmer()
stems = [stemmer.stem(w) for w in words_no_sw]

print(stems)
```

```python
def get_wordnet_pos(treebank_tag):
  if treebank_tag.startswith('J'):
    return wordnet.ADJ
  elif treebank_tag.startswith('V'):
    return wordnet.VERB
  elif treebank_tag.startswith('N'):
    return wordnet.NOUN
  elif treebank_tag.startswith('R'):
    return wordnet.ADV
  else:
    return wordnet.NOUN

```

```python
lemmatizer = WordNetLemmatizer()
pos_no_sw = pos_tag(words_no_sw)

lemmatized_words = []

for word,tag in pos_no_sw:
  wn_tag = get_wordnet_pos(tag)
  lemma = lemmatizer.lemmatize(word,wn_tag)
  lemmatized_words.append(lemma)


print(lemmatized_words)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
#Part 2 TF-IDF

corpus = [
    "NLP is best",
    "NLP is very helpfull for AI",
    "CR7 is the goat"
]
```

```python
vectorizer = TfidfVectorizer()
```

```python
tfidf_matrix = vectorizer.fit_transform(corpus)
```

```python
terms = vectorizer.get_feature_names_out()
print(terms)
```

```python
dense_df = pd.DataFrame(tfidf_matrix.toarray())
print(dense_df)
```

```python
doc0_vec = tfidf_matrix.toarray().flatten()

for term,value in zip(terms,doc0_vec):
  if value>0 :
    print(f"{term} ->{value}")
```
